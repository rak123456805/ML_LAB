```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing

data =fetch_california_housing(as_frame=True)
df=data.frame
print("--Histogram--")
df.hist(bins=30, figsize=(10,8))
plt.tight_layout()
plt.show()

print("--Boxplot--")

df.plot(kind='box',figsize=(10,8),subplots=True,layout=(3,3))
plt.tight_layout()
plt.show()
```
